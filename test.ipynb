```python
from __future__ import annotations

from datetime import datetime, timezone
from decimal import Decimal
from typing import Any, Literal

import httpx


RENILE_DATA_URL = "https://renile-iot.com/api/v1/data/"

DataType = Literal["month_hours", "day"]


def parse_datetime(value: str) -> datetime:
    value = value.strip()

    if value.endswith("Z"):
        value = value.replace("Z", "+00:00")

    try:
        dt = datetime.fromisoformat(value)
    except ValueError:
        dt = datetime.strptime(value, "%Y-%m-%d %H:%M")

    if dt.tzinfo is not None:
        dt = dt.astimezone(timezone.utc).replace(tzinfo=None)

    return dt


def parse_decimal_value(value: Any) -> float | int | None:
    if value is None:
        return None

    if isinstance(value, dict) and "$numberDecimal" in value:
        value = value["$numberDecimal"]

    try:
        decimal_value = Decimal(str(value))
    except Exception:
        return None

    if decimal_value == decimal_value.to_integral_value():
        return int(decimal_value)

    return float(decimal_value)


def truncate_sensor_chart_response(
    response: dict[str, Any],
    start_time: str,
    end_time: str,
) -> dict[str, Any]:
    start_dt = parse_datetime(start_time)
    end_dt = parse_datetime(end_time)

    if end_dt < start_dt:
        raise ValueError("end_time must be greater than or equal to start_time")

    result: dict[str, Any] = {}

    for sensor_name, sensor_payload in response.items():
        if not isinstance(sensor_payload, dict):
            continue

        labels = sensor_payload.get("labels", [])
        data = sensor_payload.get("data", [])

        if not isinstance(labels, list) or not isinstance(data, list):
            continue

        filtered_labels: list[str] = []
        filtered_data: list[float | int | None] = []

        for label, value in zip(labels, data):
            if not isinstance(label, str):
                continue

            label_dt = parse_datetime(label)

            if start_dt <= label_dt <= end_dt:
                filtered_labels.append(label)
                filtered_data.append(parse_decimal_value(value))

        result[sensor_name] = {
            "labels": filtered_labels,
            "data": filtered_data,
            "count": len(filtered_labels),
        }

    return result


async def fetch_renile_data(
    jwt: str,
    device_id: str,
    start_time: str,
    data_type: DataType,
) -> dict[str, Any]:
    headers = {
        "Authorization": f"JWT {jwt}",
    }

    params = {
        "data_type": data_type,
        "start_time": start_time,
        "device_id": device_id,
    }

    async with httpx.AsyncClient(timeout=30.0) as client:
        response = await client.get(
            RENILE_DATA_URL,
            headers=headers,
            params=params,
        )

    if response.status_code != 200:
        raise RuntimeError(
            f"ReNile API error: {response.status_code} - {response.text}"
        )

    data = response.json()

    if not isinstance(data, dict):
        raise ValueError(f"Expected dict response, got {type(data).__name__}")

    return data


async def get_historical_readings(
    jwt: str,
    device_id: str,
    start_time: str,
    end_time: str,
) -> dict[str, Any]:
    raw_response = await fetch_renile_data(
        jwt=jwt,
        device_id=device_id,
        start_time=start_time,
        data_type="month_hours",
    )

    return truncate_sensor_chart_response(
        response=raw_response,
        start_time=start_time,
        end_time=end_time,
    )


async def get_historical_summary(
    jwt: str,
    device_id: str,
    start_time: str,
    end_time: str,
) -> dict[str, Any]:
    raw_response = await fetch_renile_data(
        jwt=jwt,
        device_id=device_id,
        start_time=start_time,
        data_type="day",
    )

    return truncate_sensor_chart_response(
        response=raw_response,
        start_time=start_time,
        end_time=end_time,
    )
    
def convert_for_llm(response: dict) -> dict:
    result = {}

    for sensor_name, sensor_data in response.items():
        labels = sensor_data.get("labels", [])
        values = sensor_data.get("data", [])

        result[sensor_name] = {
            label: value
            for label, value in zip(labels, values)
        }

    return result
```

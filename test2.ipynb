```python
from typing import Any


def simplify_projects_devices(response: list[dict[str, Any]]) -> dict[str, Any]:
    if not response:
        return {
            "project_name": None,
            "devices": []
        }

    project_name = (
        response[0]
        .get("_project", {})
        .get("type")
    )

    devices = []

    for device in response:
        device_name = device.get("name")
        device_id = device.get("id") or device.get("_id")

        if not device_name or not device_id:
            continue

        devices.append({
            "device_name": device_name,
            "device_id": device_id
        })

    return {
        "project_name": project_name,
        "devices": devices
    }

simplify_projects_devices(response=response)
```
